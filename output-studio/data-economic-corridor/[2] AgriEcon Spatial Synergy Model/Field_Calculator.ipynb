```python
#pembuatan klasifikasi kemiringan

import pandas as pd
import ipywidgets as widgets
from IPython.display import display
from google.colab import files

print("Upload CSV files")
uploaded = files.upload()

filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)

if "Slope" not in df.columns:
  raise ValueError("Slope column not found in datasets.")


def reclassify_slope(slope):
  if -2 <= slope < 134:
    return 'S1'
  elif 134 <= slope < 382:
    return 'S2'
  elif 382 <= slope < 569:
    return 'S3'
  else:
    return 'N'


df['Class_slope'] = df['Slope'].apply(reclassify_slope)

print("Data setelah reklasifikasi:")
display(df.head(10))

output_filename = "reclassified_slope.csv"
df.to_csv(output_filename, index=False)
print(f"File saved as: {output_filename}")

files.download(output_filename)
```

```python
#klasifikasi pertama
import pandas as pd
import ipywidgets as widgets
from IPython.display import display
from google.colab import files

# 1️⃣ Upload File CSV
print("Upload file CSV yang berisi data Class_rain dan Class_soil")
uploaded = files.upload()

# 2️⃣ Membaca file yang diupload
filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)

# 3️⃣ Pastikan ada kolom yang dibutuhkan
if "Class_rain" not in df.columns or "Class_soil" not in df.columns:
    raise ValueError("Kolom 'Class_rain' atau 'Class_soil' tidak ditemukan dalam dataset!")

# 4️⃣ Urutan Kelas dari Tinggi ke Rendah
class_rank = {'S1': 1, 'S2': 2, 'S3': 3, 'N': 4}  # 1 = terbaik, 4 = terburuk

# 5️⃣ Fungsi untuk Menentukan Kelas Akhir
def determine_final_class(rain, soil):
    if rain == soil:
        return rain  # Jika sama, ambil salah satu (karena nilainya sama)
    else:
        return min(rain, soil, key=lambda x: class_rank[x])  # Ambil kelas terbaik (ranking terendah)

# 6️⃣ Terapkan Fungsi ke DataFrame
df['Final_Class'] = df.apply(lambda row: determine_final_class(row['Class_rain'], row['Class_soil']), axis=1)

# 7️⃣ Menampilkan hasil
print("Data setelah pengolahan:")
display(df.head(10))  # Menampilkan 10 baris pertama

# 8️⃣ Simpan hasil ke file CSV
output_filename = "final_classification.csv"
df.to_csv(output_filename, index=False)
print(f"File telah disimpan sebagai: {output_filename}")

# 9️⃣ Unduh hasilnya
files.download(output_filename)
```

```python
#pembuatan pembobotan_union_klasifikasi

import pandas as pd
from google.colab import files

# Upload File CSV
print("Upload file CSV yang berisi kolom 'Class_rain' dan 'Class_soil'")
uploaded = files.upload()

# Membaca file yang diupload
filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)

# Pastikan kolom yang dibutuhkan ada
if "Class_slope" not in df.columns or "Final_Class" not in df.columns:
    raise ValueError("Kolom 'Class_slopel' atau 'Final_Class' tidak ditemukan dalam dataset!")

# Urutan Kelas (untuk menentukan tingkat kesesuaian)
kelas_hierarki = {'S1': 1, 'S2': 2, 'S3': 3, 'N': 4, 'No Data':5} #Added 'No Data' as a key

# Fungsi untuk Menentukan Kelas Terendah
def get_lowest_class(final, slope):
    if final == slope:
        return final  # Jika sama, ambil salah satu

    # Handle cases where rain or soil is 'No Data'
    if final == 'No Data' or slope == 'No Data':
      return 'No Data'

    #Handle cases for S1, S2, S3, and N
    return final if kelas_hierarki[final] > kelas_hierarki[slope] else slope  # Ambil kelas terendah


# Proses Reklasifikasi
df['yhat'] = df.apply(lambda row: get_lowest_class(row['Final_Class'], row['Class_slope']), axis=1)

# Menampilkan hasil reklasifikasi
print("Data setelah klasifikasi:")
display(df.head(10))

# Simpan hasil ke file CSV
output_filename = "Classification_fin.csv"
df.to_csv(output_filename, index=False)
print(f"File telah disimpan sebagai: {output_filename}")

# Unduh hasilnya
files.download(output_filename)
```
